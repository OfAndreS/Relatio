```python
df_d07_clean = df_d07.copy()
```

##### 1.1 ) _Renomeando colunas_

```python
# Aplicando a transformação na variável "fnlwgt"

df_d07_clean = df_d07_clean.rename(columns={'fnlwgt' : 'final_weight'})
df_d07_clean.columns.to_list()[2]
```

##### 1.2 ) _Transformando variáveis categóricas em binárias_

```python
# Aplicando a transformação na variável "income"

map_income = {'<=50K' : 0, '>50K' : 1}
df_d07_clean['income'] = df_d07['income'].map(map_income)
df_d07_clean['income'].unique()
```
