#### **02 ) - Análise da correlação Numérico / Numérico**

- 

---

```python
df_d07_num = df_d07_status[(df_d07_status['variable_type'] == 'quantitative') & (df_d07_status['variable_subtype'] == 'continuous')][['column', 'variable_type','variable_subtype']].reset_index(drop=True)
df_d07_num
```

```python
df_d07_num_pearson = df_d07_clean[df_d07_num['column'].values.tolist()].corr(method='pearson')
df_d07_num_pearson
```

```python
top_10_d07_num_pearson = (
    df_d07_num_pearson.unstack().reset_index(name='Valor').set_axis(['Coluna', 'Linha', 'Valor'], axis=1).drop_duplicates(subset=['Valor']).query('Valor != 1').nlargest(10, 'Valor')
)

top_10_d07_num_pearson.index = range(1, 11)
top_10_d07_num_pearson
```

```python
df_d07_num_spearman = df_d07_clean[df_d07_num['column'].values.tolist()].corr(method='spearman')
df_d07_num_spearman
```

```python
top_10_d07_num_spearman = (
    df_d07_num_spearman.unstack().reset_index(name='Valor').set_axis(['Coluna', 'Linha', 'Valor'], axis=1).drop_duplicates(subset=['Valor']).query('Valor != 1').nlargest(10, 'Valor')
)

top_10_d07_num_spearman.index = range(1, 11)
top_10_d07_num_spearman
```

```python
df_d07_num_kendall = df_d07_clean[df_d07_num['column'].values.tolist()].corr(method='kendall')
df_d07_num_kendall
```

```python
top_10_d07_num_kendall = (
    df_d07_num_kendall.unstack().reset_index(name='Valor').set_axis(['Coluna', 'Linha', 'Valor'], axis=1).drop_duplicates(subset=['Valor']).query('Valor != 1').nlargest(10, 'Valor')
)

top_10_d07_num_kendall.index = range(1, 11)
top_10_d07_num_kendall
```
