##### A ) _Funções_

```python
def load_arff_to_dataframe(dataset_num, base_path, final_path="/dataset"):
    """
    Reads an ARFF file and converts its data and columns
    into a pandas DataFrame.
    """
    file_path = base_path + str(dataset_num) + final_path
    
    with open(file_path, 'r') as f:
        arff_dict = arff.load(f)
        
    columns = [attribute[0] for attribute in arff_dict['attributes']]
    df = pd.DataFrame(arff_dict['data'], columns=columns)
    
    return df
```

```python
import math
import random

def selecionar_amostra_representativa(lista_variaveis: list, tamanho_amostra: int = 80) -> list:
    total_elementos = len(lista_variaveis)
    
    if total_elementos < tamanho_amostra:
        raise ValueError("O tamanho da lista original é inferior à amostra solicitada.")
    
    salto = total_elementos / tamanho_amostra
    amostra_espalhada = []
    
    for i in range(tamanho_amostra):
        indice_base = i * salto
        
        indice_aleatorio_no_bloco = math.floor(indice_base + random.uniform(0, salto))
        
        indice_final = min(indice_aleatorio_no_bloco, total_elementos - 1)
        
        amostra_espalhada.append(lista_variaveis[indice_final])
        
    return amostra_espalhada
```
